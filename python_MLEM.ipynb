```python
import numpy as np
import matplotlib.pyplot as plt

from skimage.data import shepp_logan_phantom
from skimage.transform import radon, rescale, iradon
```

```python
N = 128
phantom_img = shepp_logan_phantom()
```

```python
plt.imshow(phantom_img)
```

```python
#generating simulated projection data -- sinogram
theta = [i for i in range(0,179)]# projection angles
```

```python
result = radon(phantom_img,theta);
```

```python
#MLEM variables
num_iters=30 #iterations
mlem_recon=np.ones_like((N,N)) #for reconstructed image
ones_img=np.ones((N,N)) #to calculate the normalization factor
```

```python
iradon()
```

```python
radon_img = radon(ones_img, theta)
```

```python
#normalization factor (backprojection of a sinogram of all ones) -- sensitivity matrix components \sum_i a_{ij}
sens_matrix = iradon(radon_img, theta, 128)
mask = sens_matrix <=0
```
